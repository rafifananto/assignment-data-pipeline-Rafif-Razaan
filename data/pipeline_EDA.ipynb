```python
# Mengubah setting date
df['transaction_date']=pd.to_datetime(df['transaction_date'], format='mixed')
```

```python
# Mengubah setting alphabet menjadi huruf kecil semua
df['make']=df['make'].str.lower()
```

```python
# Menghilangkan spasi additional
df['make']=df['make'].str.strip()
```

```python
df['make'].unique()
```

```python
df['fuel-system']=df['fuel-system'].str.lower()
```

```python
df['fuel-system'].unique()
```

```python
# Pisahkan bulan dan tahun pada kolom "transaction-date"
df['transaction_month']=df['transaction_date'].dt.month
df['transaction_year']=df['transaction_date'].dt.year
df.drop('transaction_date', axis=1, inplace=True)
df
```

```python
from sklearn.preprocessing import MinMaxScaler, OrdinalEncoder
```

```python
from sklearn.preprocessing import MinMaxScaler, OrdinalEncoder

# 1. Ordinal Encoding: Untuk 'horsepower-binned'
# Asumsi urutan kategorinya adalah Low, Medium, High
ordinal_cols = ['horsepower-binned']
ord_encoder = OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
df[ordinal_cols] = ord_encoder.fit_transform(df[ordinal_cols])

# 2. One-Hot Encoding: Menggunakan pandas get_dummies agar praktis
nominal_cols = ['make', 'aspiration', 'body-style', 'drive-wheels', 
                'engine-location', 'engine-type', 'fuel-system']
df = pd.get_dummies(df, columns=nominal_cols, drop_first=True, dtype=int)

# 3. Min-Max Scaling: Untuk semua kolom numerik kontinu
numeric_cols = ['normalized-losses', 'wheel-base', 'length', 'width', 'height', 
                'curb-weight', 'engine-size', 'bore', 'stroke', 'compression-ratio', 
                'horsepower', 'peak-rpm', 'city-mpg', 'highway-mpg', 'city-L/100km']
                
scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

# Cek hasil akhir
print(df.head())
```
